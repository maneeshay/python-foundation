```python
# If statement :
# check if the number is positive
num = 55
if num > 0:
    print("given number is positive")
```

```python
# Elif
a = 33
b = 33
if a > b:
    print("a is greater")
elif a < b:
    print("b is greater")
else:
    print("both are equal")
```

```python
# categorizing the age group:
age = 25
if age < 13:
    print("You are child")
elif age <20:
    print("you are teenager")
elif age < 65:
    print("you are an adult")
else:
    print("you are senior")
```

```python
# Nested If
x = 42
if x > 10:
    print("Above 10")
    if x > 20:
        print("Above 20")
    else:
        print("Not aove 20")
        
```

## Match


```python
day = 4
match day:
    case 1:
        print("Sunday")
    case 2:
        print("Monday")
    case 3:
        print("Tuesday")
    case 4:
        print("Wednesday")
    case 5:
        print("Thursday")
    case 6:
        print("Friday")
    case _:  #default value
        print("Friday")
```

## While Loop

```python
i = 1
while i < 6:
    print(i)
    i += 1
```

```python
# with continue statement , we can stop the current iteration and continue with the next
i = 1
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

## For loops

```python
fruits = ["apple", "banana","melon"]
for fruit in fruits:
    print(fruit)
```

```python
for x in "banana":
    print(x)
```

```python
for x in range(2,20,2):
    print(x)
```

```python
# Create the fruits list
fruits = ["apple", "banana", "cherry"]

# Loop through fruits, break at "banana"
for fruit in fruits:
    print(fruit)
    if fruit == "banana":
        break
```
