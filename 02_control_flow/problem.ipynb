🧠 Problem 1 — Even or Odd

Take a number from user and print:

"Even" if number is even

"Odd" if number is odd

👉 Use:

input

int

if

Write the code yourself first.

```python
num = int(input("Enter any number: "))
if num % 2 == 0:
    print("Given number is Even")         
else:
    print("It's odd")
          
```

🧠 Problem 3 — Sum of First N Numbers

Ask user for a number n.

If user enters 5 → output should be:

1 + 2 + 3 + 4 + 5 = 15

Use:

loop

accumulator variable

```python
num = int(input("Enter a number: "))
total = 0
for n in range(1,num+1):
    total += n
print("total of the sum of",num,"is: ",total)
```

🧠 Problem 4 — Count Vowels in a String

Take string input.

Count how many vowels it contains.

Example:

"manisha"
→ 3

Use:

for

if

string membership (in)

```python
word = input("Ente a word: ")
count = 0
v = "aeiouAEIOU"
for letter in word:
    #print(letter)
    if letter in v:
        count += 1
        print(letter)
print("total number of vowels in ",word, "is: ",count)
```

```python
word = input("Ente a word: ")
count = 0
v = "aeiouAEIOU"
for letter in word:
    #print(letter)
    if letter in ('a','e','i','o','u'):
        count += 1
        print(letter)
print("total number of vowels in ",word, "is: ",count)
```

🧠 Problem 5 — Reverse a String (Without Slicing)

You once struggled with this — so we fix it properly.

Input: "hello"
Output: "olleh"

Use:

loop

no [::-1]

```python
word = input("Enter a word")
reverse =""
for char in word:
    #print(char)
    reverse = char + reverse
    #print(reverse)
print("Reverse of the given word is: ",reverse)
```

🧠 Problem 6 — Guess the Number

Set a fixed number:

secret = 7

Ask user to guess.

Use while loop until they guess correctly.

Print:

"Too high"

"Too low"

"Correct"

```python
secret = 7

while True:
    num = int(input("Enter the secret number"))
    if num < secret:
        print("Too low")
    elif num > secret:
        print ("too high")
    else:
        print("accurate")
        break
```
