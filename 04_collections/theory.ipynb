 Q.no. 1 List → Set → List

Given a list:

nums = [1,2,2,3,4,4,5]

Remove duplicates without using set() directly.

(You may use a loop.)

```python
nums = [1,2,2,3,4,4,45]
num1=[]
for i in nums:
    if i not in num1:
        num1.append(i)
print(num1)
```

```python
nums = [1,2,2,3,4,4,45]
seen = set()
result = []

for i in nums:
    if i not in seen:
        seen.add(i)
        result.append(i)

print(result)
```

Q.no.2 Count Frequency (Manual)

Given:

s = "banana"

Return a dictionary of character frequencies.

Expected:

{'b':1, 'a':3, 'n':2}

No Counter. Manual logic.

```python
seen = set()

for i in nums:
    seen.add(i)

print(seen)
```

```python
s ="banana"
freq ={}

for char in s:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print(freq)
```

```python
s = "swiss"
freq={}

# build a frequency table
for char in s:
    freq[char] = freq.get(char,0)+1
print(freq)
 
#frind the 1st non-repeating char
for char in s:
    if freq[char] == 1:
        print(char)
        break
    
```
