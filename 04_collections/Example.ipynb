## list

```python
# are orderd, changable, and allow duplicate values

thelist = ["apple", "banana", "mango"]
print(len(thelist))
```

```python
# indexing
fruits = ["apple","banana","kiwi","mango"]
print(fruits[0])


#negative indexing
print(fruits[-1])
```

```python
# range of indexing
thislist = ["apple", "banana", "cherry", "orange","kiwi","apple"]
print(thislist[2:5])

print(thislist[:4])
print(thislist[:])

#negative indexing
print(thislist[-4:-1])
```

```python
#check if the item exist 
thislist=["apple","banana","mango","cherry"]
if "apple" in thislist:
    print("true")
```

```python
#change the item in the list: 
fruits =["apple","banana","cherry","kiwi"]
fruits[1:2] = ["mango","grape"]
print(fruits)

#insert
fruits[2] = "watermelon"
print(fruits)
```

```python
 #Create a list
colors = ["red","black","green"]
print(colors)
# Print the first item
print(colors[0])
# Change the second item to "yellow"
colors[1] = "yellow"
print(colors)
# Add "purple" to the end
colors.append("purple")
# Remove "red"
colors.remove("red")
# Print the list
print(colors)
```

## TUPLE

```python
# Tuples is collection which is orderd and unchangeable
mytuple = ("mango","apple", "banana")
print(mytuple)
```

```python
# to create a tuple with only one item we have to adda come,
thistuple = ("apple",)
print(thistuple)
```

```python
thistuple = tuple((1,23,3,4))
print(type(thistuple))

#access the element
print(thistuple[1])

#everything else almost like list
```

```python
# check if the item exist
fruits =("apple","banana","cherry")
if "apple" in fruits:
    print("yes the fruit exist in the tuple")
```

```python
# since the tuple is unchagable, but there's loop if you want to change the value, we change the tuple into list
# change it's value and change the tuple back into tuple

x = ("apple", "banana","kiwi")
y = list(x)
y[1] = "mango"
y.append("melon")
x = tuple(y)
print(x)
```

```python
# Add a tuple to tuple
x = ("apple","orange","cherry")
y = ("kiwi","melon")
x += y
print(x)
```

```python
x = ("apple","orange","cherry")
print(x)
del x
print(x)
```

```python
# Unpacking a tuple
fruits = ("apple","banana","mango")
(apple,mango,banana) = fruits
print(apple)
```

```python
#Using asterick
fruits =("apple","banana","cherry","strawberry","raspberry")
(green,yellow,*red) = fruits
print(green)
print(yellow)
print(red)  #all the remaining value will assign to it as a list
```

```python
#Using asterick
fruits =("apple","banana","cherry","strawberry","raspberry")
(green,*yellow,red) = fruits
print(green)
print(yellow)
print(red)  
```

```python
# Loop through the toople 
x = ("white", "green","red")
for i in x:
    print(i)
```

```python
x = ("white", "green","red")
for i in range(len(x)):
    print(x[i])
```

```python
# Create the tuple
fruits = ("apple","banana","cherry")
# Print the second item
print(fruits[1])
# Print the number of items
print(len(fruits))
# Unpack the tuple
(a,b,c) = fruits
# Print b
print(b)
```

## Dictionaries

* used to store datavalues in key:value pairs
* it's ordered, changable and do not allow duplicate value

```python
thisdict ={
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)
```

```python
thisdict ={
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
    "year": 2026    #duplicate value overwrite the existing value
}
print(thisdict)
```

```python
thisdict = dict(name = "john", age = 24, country = "USA")
print(thisdict)
```

```python
thisdict ={
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)

#accessing the element
x = thisdict["model"]
print(x)

print(thisdict.keys())
print(thisdict.values())
print(thisdict.items())
```

```python
car ={
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

print(car)

car["color"] = "black" #add the the value
print(car)

if "brand" in car:
    print("the brand of the car exists")
```

```python
car ={
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

car.update({"year":2020})
print(car)

for x in car.values():
    print(x)
print("--------")   
for x in car.keys():
    print(x)
print("--------")
for x in car:
    print(x)
print("--------")
for x in car.items():
    print(x)
```
